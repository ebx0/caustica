# caustica GPU gate suite

This notebook is **project maintenance**, not a user flow: it runs caustica's own
milestone gates against a real GPU and writes a stamped PASS/FAIL report. If you want to
run a simulation, use [`colab_run.ipynb`](https://github.com/ebx0/caustica/blob/master/notebooks/colab_run.ipynb) instead.

Before you start: **Runtime -> Change runtime type -> Hardware accelerator: GPU.** A CPU
runtime exits with the fix and prepares nothing.

What it measures (M7 and M8's on-device criteria, in one run): an on-device calibration, a
VRAM ladder of increasing jobs with predicted-vs-measured memory and wall time, one rung
that is deliberately too large and must be REFUSED, and a numpy-vs-cupy field comparison.
Every scenario is generated from the library — homogeneous water plus a bowl scaled to the
grid — so nothing is downloaded and nothing external is needed.

Expect **10-20 minutes** on an A100 and several GB of host RAM (the largest rungs build a
property volume before the plan-first gate can even look at them). Exit code: 0 every gate
passed, 2 no usable GPU, 4 a gate failed or is still incomplete.
